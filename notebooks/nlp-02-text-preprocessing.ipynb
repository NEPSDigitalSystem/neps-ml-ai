# Text Preprocessing

### Objective

Clean and prepare data for ML

```python
import pandas as pd

df = pd.read_csv("C:/Users/yasmine/Documents/neps/neps-ml-ai/data/raw/nlp_responses.csv")
```

#### Clean Data

```python
import re

def clean_text(text):
    
    text = str(text).lower()

    text = re.sub(r"http\S+", "", text)
    text = re.sub(r"[^a-zA-Z\s]", "", text)
    text = re.sub(r"\s+", " ", text).strip()

    return text


df["clean_text"] = df["response_text"].apply(clean_text)
```

#### Tokenization

```python
import nltk
nltk.download("punkt")

from nltk.tokenize import word_tokenize

df["tokens"] = df["clean_text"].apply(word_tokenize)
```

#### Stop word remover

```python
nltk.download("stopwords")
from nltk.corpus import stopwords

stop_words = set(stopwords.words("english"))

df["tokens"] = df["tokens"].apply(
    lambda x: [word for word in x if word not in stop_words]
)
```

#### Lemmatization

```python
nltk.download("wordnet")

from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

df["tokens"] = df["tokens"].apply(
    lambda x: [lemmatizer.lemmatize(word) for word in x]
)
```

```python
df["clean_text_final"] = df["tokens"].apply(
    lambda x: " ".join(x)
)
```

#### Save data

```python
df.to_csv(
    "C:\\Users\\yasmine\\Documents\\neps\\neps-ml-ai\\data\\preprocessed\\clean_nlp_data.csv",
    index=False
)
```
