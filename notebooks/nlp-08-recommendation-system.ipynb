```python
def recommend(row):

    if row['severity_level']=="high":
        return "Urgent referral"

    elif row['stress_level']=="high":
        return "Stress counselling"

    elif row['anxiety_level']=="high":
        return "CBT intervention"

    else:
        return "Routine monitoring"
```

```python
df["recommendation"] = df.apply(
    recommend,
    axis=1
)
```
